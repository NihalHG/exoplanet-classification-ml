## Inspecting Possible Outcome Labels

The unique values and frequency counts of the possible outcome columns will be inspected before selecting a target variable.

The `koi_disposition` column contains three observed labels: `FALSE POSITIVE`, `CONFIRMED`, and `CANDIDATE`.

The `koi_pdisposition` column contains two observed labels: `FALSE POSITIVE` and `CANDIDATE`.

These columns appear to represent different outcome definitions or stages. Their exact meanings must be verified using the NASA Kepler Objects of Interest documentation before selecting a target variable.

Because both columns are outcome-related, they will be treated as possible sources of target leakage and will not automatically be used as input features.

## Checking Missing Values

Missing values can affect summary statistics, feature selection, and later model training.

The number of missing values in each column will be inspected before deciding which columns can be used. Missing-value handling will be based on the meaning and role of each column rather than applying one automatic rule to the entire dataset.

```python
print("Checking for missing values in the dataset:")
missing_values = df.isnull().sum()
print(missing_values[missing_values > 0].sort_values(ascending=False))
```

## Identifying Completely Empty Columns

Some columns contain missing values for every row, while others are only partially missing.

Columns that are completely empty cannot provide information for exploratory analysis or model training. They will be identified separately from partially missing columns. No columns will be removed until their names and meanings have been reviewed.

```python
comletely_empty_cols = df.columns[df.isnull().all()]
print("Columns that are completely empty:", comletely_empty_cols.tolist())
```

## Removing Completely Empty Columns

Three columns contain no observed values in any row. Since completely empty columns cannot contribute information to exploratory analysis or model training, they are excluded from a separate working DataFrame.

The original DataFrame is preserved so that preprocessing decisions remain transparent and reversible. Partially missing columns are not removed automatically.

```python
df_clean=df.dropna(axis=1, how="all")
print("Shape of the dataset before cleaning:", df.shape)
print("Shape of the cleaned dataset:", df_clean.shape)
print("Columns that were removed due to being completely empty:")
print([col for col in df.columns if df[col].isnull().all() and col not in df_clean.columns])
```
